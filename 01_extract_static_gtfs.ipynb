# Download & Extract Source Zip Files

```python
import requests
regular_gtfs_download = "https://rrgtfsfeeds.s3.amazonaws.com/gtfs_subway.zip"
supplemented_gtfs_download = "https://rrgtfsfeeds.s3.amazonaws.com/gtfs_supplemented.zip"
regular_gtfs_file_path = "/Volumes/nyc_subway_data/default/subway_schedule/regular_gtfs.zip"
supplemented_gtfs_file_path = "/Volumes/nyc_subway_data/default/subway_schedule/supplemented_gtfs.zip"

with requests.get(regular_gtfs_download, stream=True, timeout=10) as r:
  r.raise_for_status()
  with open(regular_gtfs_file_path, 'wb') as f:
    for chunk in r.iter_content(chunk_size=8192): 
      f.write(chunk)

with requests.get(supplemented_gtfs_download, stream=True, timeout=10) as r:
  r.raise_for_status()
  with open(supplemented_gtfs_file_path, 'wb') as f:
    for chunk in r.iter_content(chunk_size=8192): 
      f.write(chunk)


```

```python
import zipfile

with zipfile.ZipFile("/Volumes/nyc_subway_data/default/subway_schedule/regular_gtfs.zip", 'r') as zip_ref:
    zip_ref.extractall("/Volumes/nyc_subway_data/default/subway_schedule/regular/")

with zipfile.ZipFile("/Volumes/nyc_subway_data/default/subway_schedule/supplemented_gtfs.zip", 'r') as zip_ref:
    zip_ref.extractall("/Volumes/nyc_subway_data/default/subway_schedule/supplemented/")
```
