# Create Reusable Loading Notebook

```python
file_path = dbutils.widgets.get("file_path")

df = (spark.read
      .format("csv")
      .option("header", "true")
      .option("delimiter", ",")
      .load(file_path))
```

```python
display(df)
df.printSchema()
```

```python
from pathlib import Path

subject = Path(file_path).stem
feed_type = Path(file_path).parent.name
table_name = f"bronze_{subject}_{feed_type}"

(df.write
 .format("delta")
 .mode("overwrite")
 .saveAsTable(f"nyc_subway_data.default.{table_name}")
)
```
